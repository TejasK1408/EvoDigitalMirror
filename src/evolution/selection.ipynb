```python
import time
import random
# Six dresses in our shopping bucket

population = [
    {"id": 1, "image": "dress1.jfif", "dwell_time": 0.0},
    {"id": 2, "image": "dress2.jfif", "dwell_time": 0.0},
    {"id": 3, "image": "dress3.jfif", "dwell_time": 0.0},
    {"id": 4, "image": "dress4.jfif", "dwell_time": 0.0},
    {"id": 5, "image": "dress5.jfif", "dwell_time": 0.0},
    {"id": 6, "image": "dress6.jfif", "dwell_time": 0.0}
]

print("Initial population created!")

for dress in population:
    print(dress)
    
def update_dwell_time(dress_id, elapsed_time):

    if dress_id is None:
        return

    for dress in population:

        if dress["id"] == dress_id:
            dress["dwell_time"] += elapsed_time
            break
        
def select_best_two():

    ranked_population = sorted(
        population,
        key=lambda dress: dress["dwell_time"],
        reverse=True
    )

    best_two = ranked_population[:2]

    return best_two

# Simulated dwell times (seconds)

simulated_dwell_times = {
    1: 4.8,
    2: 3.9,
    3: 1.2,
    4: 0.8,
    5: 2.1,
    6: 0.5
}

for dress_id, dwell in simulated_dwell_times.items():
    update_dwell_time(dress_id, dwell)


# Automatically select the two best dresses

parents = select_best_two()

print("\n========== SELECTED DRESSES ==========")

for dress in parents:

    print(
        f"Dress {dress['id']} | "
        f"Image: {dress['image']} | "
        f"Dwell time: {dress['dwell_time']:.2f}s"
    )
    
import matplotlib.pyplot as plt

dress_ids = [dress["id"] for dress in population]
dwell_times = [dress["dwell_time"] for dress in population]

plt.figure(figsize=(9, 5))

bars = plt.bar(dress_ids, dwell_times)

plt.xlabel("Dress ID")
plt.ylabel("Dwell Time (seconds)")
plt.title("Gaze-based Dress Selection")

plt.xticks(dress_ids)
plt.grid(axis="y", alpha=0.3)

for dress, bar in zip(population, bars):

    if dress in parents:
        bar.set_color("green")

plt.show()
```
